```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (accuracy_score,confusion_matrix,
classification_report)
```

```python
dataset = pd.read_csv("onlinefraud.csv")
```

```python
sample_size = min(100000, len(dataset))
dataset = dataset.sample(n=sample_size, random_state=42)
```

```python
print("First Five Records")
print(dataset.head())

print("\nDataset Shape")
print(dataset.shape)

print("\nDataset Information")
dataset.info()
```

```python
print("\nMissing Values")
print(dataset.isnull().sum())
```

```python
FEATURE_COLUMNS = [
    "step",
    "amount",
    "oldbalanceOrg",
    "newbalanceOrig",
    "oldbalanceDest",
    "newbalanceDest"
]

TARGET_COLUMN = "isFraud"

required_columns = FEATURE_COLUMNS + [TARGET_COLUMN]

missing_columns = [
    column
    for column in required_columns
    if column not in dataset.columns]

if missing_columns:
    raise ValueError(
        f"Missing columns in dataset: {missing_columns}")
```

```python
dataset = dataset.dropna(subset=required_columns)
```

```python
X = dataset[FEATURE_COLUMNS]

y = dataset[TARGET_COLUMN]

print("\nTarget Distribution")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,
test_size=0.20,random_state=42,stratify=y)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1)

model.fit(X_train,y_train)

print("\nRandom Forest Model Trained Successfully")
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)

print("\nAccuracy")
print(accuracy)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred)

print("\nConfusion Matrix")
print(cm)
```

```python
print("\nClassification Report")

print(classification_report(y_test,y_pred,zero_division=0))
```

```python
sample = pd.DataFrame(
    [[
        1,
        5000,
        10000,
        5000,
        0,
        5000
    ]],
    columns=FEATURE_COLUMNS)
```

```python
prediction = model.predict(sample)

print("\nSample Transaction")
print(sample)

if prediction[0] == 1:
    print("\nPrediction: Fraud Transaction")
else:
    print("\nPrediction: Legitimate Transaction")
```

```python
probability = model.predict_proba(sample)

print(
    "\nFraud Probability:",
    probability[0][1])

print(
    "Legitimate Probability:",
    probability[0][0])
```
